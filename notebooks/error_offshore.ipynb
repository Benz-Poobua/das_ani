### Offshore Comparison

````python
import glob
import sys

sys.path.append('..')

from src.plot import animate_ncf_comparison, animate_fv_ssim_tracking
from src.ncf import get_vs_number
from src.error import ssim_index

import matplotlib as mpl
mpl.rcParams["animation.html"] = "jshtml"
mpl.rcParams["animation.embed_limit"] = 300
````

````python
# Choose dataset
pattern_conventional_1d_s1 = "../data/ncf_pre_offshore/20180819_cc_*_1d_conventional_s1.npz"
pattern_conventional_1d_s2 = "../data/ncf_pre_offshore/20180819_cc_*_1d_conventional_s2.npz"

pattern_v1_1d_s1 = "../data/ncf_pre_offshore/20180819_cc_*_1d_v1_s1.npz"
pattern_v1_1d_s2 = "../data/ncf_pre_offshore/20180819_cc_*_1d_v1_s2.npz"

files_s1_conventional = sorted(glob.glob(pattern_conventional_1d_s1), key=get_vs_number)
files_s2_conventional = sorted(glob.glob(pattern_conventional_1d_s2), key=get_vs_number)

files_s1_v1 = sorted(glob.glob(pattern_v1_1d_s1), key=get_vs_number)
files_s2_v1 = sorted(glob.glob(pattern_v1_1d_s2), key=get_vs_number)

print("Matched files (s1):", len(files_s1_conventional))
print("Matched files (s2):", len(files_s2_conventional))
print("First 3 (s1):", files_s1_conventional[:3])
print("First 3 (s2):", files_s2_conventional[:3])
````

#### 1. NCF Errors

````python
ani_comparison = animate_ncf_comparison(
    files_A=files_s1_conventional[:20],  
    files_B=files_s1_v1[:20],             
    label_A="Conventional",
    label_B="v1",
    unit="km",
    pclip=99.0,           
    pclip_diff=99.0,     
    range_m=3000.0,             
    view_side="both",          
    pos_offset=0.0,
    interval_ms=300, 
    save_vs=[930]
)

ani_comparison
````

#### 2. SSIM

````python
# 1. Grab and sort everything once
fv_out_dir = "../results/fv_panels_offshore"
all_saved_fv = sorted(glob.glob(f"{fv_out_dir}/*_fv.npz"), key=get_vs_number)

# 2. Filter out two specific subsets
files_conventional_s1 = [
    f for f in all_saved_fv 
    if "_right" in f and "_s1" in f and "conventional" in f
]

files_v1_s1 = [
    f for f in all_saved_fv 
    if "_right" in f and "_s1" in f and "v1" in f
]

# 3. Generate the 3-Panel SSIM Tracker
ani_ssim = animate_fv_ssim_tracking(
    files_A=files_conventional_s1[:30],
    files_B=files_v1_s1[:30],
    ssim_func=ssim_index, 
    label_A="Conventional",
    label_B="v1",
    xmin=0.4, xmax=2.0,
    ymin=200.0, ymax=1500.0,
    cmap="viridis",
    interval_ms=300,
    p=99.5,
    gaussian_weights=True,
    sigma=1.5, 
    save_vs=[930]
)

ani_ssim
````
